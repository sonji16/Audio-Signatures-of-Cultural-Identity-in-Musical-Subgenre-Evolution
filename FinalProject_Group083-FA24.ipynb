# Hypothesis


We hypothesize that when music genres are adopted by different ethnic groups, they will display statistically significant differences in measurable audio characteristics (i.e. tempo, energy, instrumentalness), with each ethnic group showing consistent patterns on how they modify these features. More tangibly, we predict that the subgenres will exhibit quantifiable variations in their audio measurements that correlate with the ethnic background of the artists.

This hypothesis is based on several factors. Firstly, cultural differences in musical traditions (i.e. rhythmic patterns, scale systems, instrumental preferences) are very likely to influence how artists adapt and interpret genres. Secondly, geographic/linguistic variations between ethnic groups often lead to particularly unique musical expressions, even within the same “framework” of genre (i.e. how tempo and rhythm might be modified to better accommodate different language patterns, or how traditional instruments might be incorporated into modern genres). Our hypothesis is also informed by empirical observations of how genres like hip-hop, pop, and electronic music have evolved differently across various cultural contexts, suggesting systematic patterns in how musical elements are modified to align with local musical traditions.

We first import our necessary dependencies:

````python
import sqlite3
import pandas as pd
import numpy as np
import json
import matplotlib.pyplot as plt
import seaborn as sns
import plotly.express as px
from plotly.subplots import make_subplots
import plotly.graph_objects as go
from scipy.stats import kruskal
````

We can establish a "connection" to our SQLite database using the `sqlite3` library with `.connect()`, and load the `FEATURES_2` and `TRAINING_DB_2` tables into pandas DataFrames with `.read_sql_query()`:

````python
connection = sqlite3.connect("data/last_fm_sample.db")

features_df = pd.read_sql_query("SELECT * FROM FEATURES_2", connection)
training_df = pd.read_sql_query("SELECT * FROM TRAINING_DB_2", connection)

display(features_df.head())
display(training_df.head())
````

We need to merge these two tables on `track_id` to get a single `DataFrame` with all the information we need. Since a track can have multiple genres (and thus can appear in multiple rows in the `TRAINING_DB_2` table), we'll need to aggregate the genres for each track so that each track only appears once in our merged `DataFrame`. If a track has multiple genres, they will be separated by commas in the `genre_name` column:

````python
genres_combined = (
    training_df.groupby("track_id")["genre_name"]
    .agg(lambda x: ",".join(sorted(set(x.str.strip()))))
    .reset_index()
)

genres_combined
````

We perform an inner join on `track_id` to merge the two `DataFrame`s:


````python
merged_df = pd.merge(features_df, genres_combined, on="track_id", how="inner")

# For display purposes, we're showing columns that are semantic for humans:
merged_df[["track_id", "album_name", "artist_name", "genre_name"]]
````

Following this, we take a 10% random sample of the merged `DataFrame`:


````python
np.random.seed(42)

sampled_df = merged_df.sample(frac=0.1)

# Take a look at the indices of the sampled DataFrame
sampled_df[["track_id", "album_name", "artist_name", "genre_name"]]
````

For some reason there's a bunch of genres called "Everything" that are polluting the genres column. If it is one of multiple genres for a song, we'll remove it from the cell. If it is the only genre for a song, we'll remove the song from the dataset:

````python
sampled_df["genre_name"] = sampled_df["genre_name"].str.replace("Everything", "")
sampled_df = sampled_df[sampled_df["genre_name"] != ""]
````

Finally, the most tedious part of this section is creating an empty JSON file and manually filling in genre-to-country associations. We followed a strict methodology:
- We only used countries from the `pycountry` library (which uses [ISO 3166-1](https://en.wikipedia.org/wiki/ISO_3166-1) country names)
- We only mapped genres that are associated with a particular country today. For genres that were developed in a particular country but have since expanded globally (e.g. hip-hop, reggae), we did not map them to that country.
- We excluded genres that are associated with ethnic groups that don't majority associate to an existing country, such as the Kurdish and Hmong people.

This manual mapping process helps us analyze geographic patterns in music genres while maintaining data consistency. This was the script that we used to create the JSON file:

````python
unique_genres = set()
for genres in sampled_df["genre_name"].str.split(","):
    if genres is not None:
        unique_genres.update([genre.strip() for genre in genres])

genre_country_map = {genre: "" for genre in sorted(unique_genres)}

with open("data/genre_country_mapping.json", "w", encoding="utf-8") as f:
    json.dump(genre_country_map, f, indent=4, ensure_ascii=False)
````

Adding the `country` column to our dataset:

````python
with open('data/filled_genre_country_mapping.json', 'r', encoding='utf-8') as f:
    genre_country_map = json.load(f)

def get_country_from_genres(genre_str):
    genres = genre_str.split(',')
    countries = []
    for genre in genres:
        if genre in genre_country_map and genre_country_map[genre]:
            countries.append(genre_country_map[genre])

    if not countries:
        return ''

    unique_countries = list(set(countries))

    # If more than 4 different countries, return empty string
    if len(unique_countries) > 4:
        return ''

    # If multiple countries, return most common one
    if len(unique_countries) > 1:
        return max(countries, key=countries.count)

    return countries[0]

sampled_df['country'] = sampled_df['genre_name'].apply(get_country_from_genres)

sampled_df[['album_name', 'artist_name', 'genre_name', 'country']]
````

````python
correlation_matrix = data[audio_features].corr()

corr_df = correlation_matrix.reset_index().melt(id_vars='index')
corr_df.columns = ['x', 'y', 'correlation']

fig = px.imshow(
    correlation_matrix,
    labels=dict(x="Features", y="Features", color="Correlation"),
    x=correlation_matrix.columns,
    y=correlation_matrix.columns,
    color_continuous_scale='RdBu',
    aspect='equal'
)

# Add text annotations manually
for i in range(len(correlation_matrix.columns)):
    for j in range(len(correlation_matrix.columns)):
        fig.add_annotation(
            x=correlation_matrix.columns[i],
            y=correlation_matrix.columns[j],
            text=str(round(correlation_matrix.iloc[j, i], 2)),
            showarrow=False,
            font=dict(color="black")
        )

fig.update_layout(
    title='Correlation Matrix of Audio Features',
    title_x=0.5,
    width=800,
    height=800
)

fig.show()

print("\nStrongest Correlations:")
print("-----------------------")
correlations = []
for i in range(len(audio_features)):
    for j in range(i + 1, len(audio_features)):
        corr = correlation_matrix.iloc[i, j]
        if abs(corr) > 0.3:
            correlations.append((audio_features[i], audio_features[j], corr))

for feat1, feat2, corr in sorted(correlations, key=lambda x: abs(x[2]), reverse=True):
    print(f"{feat1} vs. {feat2}: {corr:.3f}")
````

The heatmap above allows us to visualize the correlation matrix between multiple audio characteristics using colors and a pairwise correlation coefficient, $r$, which can range from -1.00 to +1.00. Positive correlations are highlighted in warm tones, and negative relationships are highlighted in cool tones. For our project, correlations exceeding the absolute coefficient value of 0.3 are considered a "strong correlation." A more red square and a higher positive coefficient value indicates a stronger positive correlation between two audio characteristics. A more blue square and a lower negative coefficient value indicates a stronger negative correlationship between two audio characteristics. A more grey square and coefficient value of 0.00 means no correlation between two audio characteristics.

As expected, diagonal audio characteristics have a correlation coefficient of +1.00 (They have a perfect correlation with themselves).

The strongest positive correlation between two audio characteristics is energy vs. loudness, which has a correlation coefficient of +0.789. This means:
- As the energy of a song increases, its loudness tends to also increase.

On the other hand, the strongest negative correlation between two audio characteristics is energy vs. acousticness, with a correlation coefficient of -0.748. This means:
- As the energy of a song increases, its acousticness tends to decrease.

There is moderately strong positive correlation between danceability vs. valence with a correlation coefficient of 0.516 and between loudness vs. valence with a correlation with a correlation of 0.310. This means:
- As the danceability of a song increases, its valence tends to increase.
- As the loudness of a song increases, its valence also tends to increase.

There is a moderately strong negative correlation between loudness vs. acousticness with a correlation coefficient of -0.608, between loudness vs. instrumentalness with a correlation coefficient of -0.366, and between instrumentalness vs. valence with a correlation coefficient of -0.313. This means:
- As the loudness of a song increases, its acousticness tends to decrease.
- As the loudness of a song increases, its instrumentalness tends to decrease.
- As the instrumentalness of a song increases, its valence tends to decrease.

Audio characteristics with near-zero correlation coefficients (e.g. speechiness and tempo) have an extremely weak linear relationship.

## Analyzing Genres Across Different Countries

### Hip-Hop/Rap

Below, we identify countries with significant hip-hop/rap presence in our dataset and analyze how different countries put their own spin on the genre through audio features. This allows us to explore how hip-hop/rap has been adapted and interpreted across different cultures:

````python
hiphop_mask = data["genre_name"].str.lower().str.contains("hip hop|rap", na=False)
hiphop_data = data[hiphop_mask].copy()

counts = hiphop_data["country"].value_counts()
selected_countries = counts[counts >= 200].index.tolist()

print("\nCountries with at least 200 hip-hop/rap tracks:")
display(selected_countries)
````

We select the five countries that each has at least 200 hip-hop/rap tracks to analyze: The United States (where hip-hop was born), Canada, Brazil, France, and Germany. These countries were chosen due to their geographical and significant cultural differences from each other, which we expect to be reflected in their audio features. To statistically validate any differences between these countries' hip-hop styles, we perform the Kruskal-Wallis H-test, which is a non-parametric test that determines whether samples from two or more groups come from the same distribution. The test computes an $H$-statistic and corresponding $p$-value. The $H$-statistic measures the extent of differences between the mean ranks of the groups—a larger $H$-statistic suggests greater differences between groups. When combined with the $p$-value (where $p < 0.05$ is considered significant), we can determine if the observed differences between countries' hip-hop styles are statistically meaningful or likely due to chance:

````python
def analyze_genre_by_country(
    data,
    countries,
    genre_regex,
    sample_size=200,
    chart_title="Audio Feature Distributions by Country",
):
    country_mask = data["country"].isin(countries)
    genre_mask = data["genre_name"].str.lower().str.contains(genre_regex, na=False)
    filtered_data = data[country_mask & genre_mask].copy()

    sampled_data = []
    for country in countries:
        country_data = filtered_data[filtered_data["country"] == country]
        if len(country_data) >= sample_size:
            sampled = country_data.sample(n=sample_size, random_state=42)
            sampled_data.append(sampled)
        else:
            print(
                f"Warning: {country} has fewer than {sample_size} tracks ({len(country_data)} tracks)"
            )
            sampled_data.append(country_data)

    filtered_data = pd.concat(sampled_data)

    fig = make_subplots(
        rows=3, cols=3, subplot_titles=[f.title() for f in audio_features]
    )

    for i, feature in enumerate(audio_features):
        row = i // 3 + 1
        col = i % 3 + 1

        box_trace = go.Box(
            x=filtered_data["country"],
            y=filtered_data[feature],
            name=feature.title(),
            showlegend=False,
        )

        fig.add_trace(box_trace, row=row, col=col)
        fig.update_xaxes(row=row, col=col)
        fig.update_yaxes(title_text=feature.title(), row=row, col=col)

    fig.update_layout(height=900, width=1200, title_text=chart_title, showlegend=False)

    fig.show()

    feature_names = []
    h_stats = []
    p_values = []
    significances = []

    for feature in audio_features:
        country_data = [
            filtered_data.loc[filtered_data["country"] == country, feature].dropna()
            for country in countries
        ]

        stat, p = kruskal(*country_data)

        feature_names.append(feature.title())
        h_stats.append(round(stat, 4))
        p_values.append(round(p, 4))
        significances.append("Significant" if p < 0.05 else "Insignificant")

    kw_results = pd.DataFrame(
        {
            "Feature": feature_names,
            "H-statistic": h_stats,
            "p-value": p_values,
            "Significance": significances,
        }
    ).set_index("Feature")

    display(kw_results)
````

````python
analyze_genre_by_country(
    data,
    selected_countries,
    "hip hop|rap",
    chart_title="Hip Hop/Rap Music Audio Feature Distributions by Country"
)
````

The bar plot graph above allows us to visualize the distributions of different audio characteristics in hip-hop tracks across our five chosen countries (the United States, Canada, Brazil, France, and Germany). Each color-coded subplot summarizes the range and median values of a certain audio characteristic (e.g tempo, dancebility, energy) for hip-hop tracks in each of the countries.

From the bar plot, we see the following trends for each audio feature:
* **Danceability:** The median dancebility is relatively similar among the five countries. However, Brazil has slightly more variability based on its interquartile range. The United States and Canada have smaller IQRs, so we can infer there is higher consistency in the danceability of hip-hop tracks from those countries.
* **Energy:** The median energy is similar among the five countries. However, Brazil has slightly greater range, and we see that some Brazilian tracks have higher energy levels. This could reflect the possibility of local influence on the hip-hop genre. We also note that Germany has one very low energy outlier which could be due to experimental style, genre fusion, or local influence.
* **Loudness:** The median loudness is similar among all countries, but Germany has the highest loudness median level while Brazil has the lowest loudness median level. Germany has the most consistent level of loudness with the smallest IQR while Brazil has more variability in loudness. Outliers with lower loudness levels are present in all five countries, with Brazil having the most significant outlier at a loudness level of -20.157.
* **Speechiness:** The United States appears to have the highest speechiness level median at 0.237 while Canada and Brazil have the lowest spechiness medians at 0.1685 and 0.164 respectively. The United States might have the highest median speechiness as it is the birthplace of hip-hop, and many American hip-hop artists place emphasis on lyricism, storytelling, and flow, or the manner in which an artist delivers their lyrics, naturally elevating speechiness levels. On the other hand, Canadian and Brazilian hip-hop may have lower median speechiness levels due to more melodic influence from regional styles of music. All five countries share similar variability in speechiness levels with their IQRs overlapping one another.
* **Instrumentalness:**  Tracks from all five countries have very low values of instrumentalness, with most close to 0, reflecting the vocal-heavy nature of hip-hop across all cultures. However, a few outliers in Brazil and Canada suggest some experimentation with more instrumental tracks.
* **Acousticness:** Acousticness levels vary drastically across the five countries. Brazil and France have the highest acoustic medians at 0.26 and 0.2455 respectively. This could possibly due to regional musical influence on the hip-hop genre. The United States has the lowest acoustic median at 0.1135.
